```python
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
```

```python
diamonds = sns.load_dataset('diamonds') #load diamons dataset
```

```python
print("First 5 rows of the dataset:")
print(diamonds.head()) #display
```

```python
print("\n--- Univariate Analysis: Carat ---")

# Statistical Summary using numpy and pandas
carat_mean = np.mean(diamonds['carat'])
carat_median = np.median(diamonds['carat'])
print(f"Mean Carat: {carat_mean:.2f}")
print(f"Median Carat: {carat_median:.2f}")

# Visualizing the distribution of the 'carat' variable
plt.figure(figsize=(8, 5))
sns.histplot(diamonds['carat'], kde=True, bins=30, color='skyblue')
plt.title('Univariate Analysis: Distribution of Diamond Carats')
plt.xlabel('Carat Weight')
plt.ylabel('Frequency')
plt.show()
```

```python
print("\n--- Multivariate Analysis: Carat vs Price ---")

# Statistical Summary: Correlation matrix between numerical variables
correlation = diamonds[['carat', 'price', 'depth', 'table']].corr()
print("Correlation Matrix:")
print(correlation)

# Visualizing relationships using a scatter plot
plt.figure(figsize=(10, 6))
sns.scatterplot(data=diamonds, x='carat', y='price', hue='cut', alpha=0.5)
plt.title('Multivariate Analysis: Carat vs Price by Cut Quality')
# Using Markdown formatting inside the plot labels for clean text rendering
plt.xlabel('Carat Weight')
plt.ylabel('Price ($)')
plt.legend(title='Cut Quality')
plt.show()
```
